```python
import os
import glob
import shutil
import re
```

```python
directory_list= './directories.txt' # path to datatool directories list
run_directory = 'v10_1_results/' # path to directory containing model run output folders
```

```python
# Read the list of paths from directories.txt
with open(directory_list, 'r') as f:
    directories = [line.strip() for line in f.readlines()]
```

```python
# Get list of files in run_directory, sort alphabetically, and exclude hidden files (starting with '.')
results_list = sorted([f for f in os.listdir(run_directory) 
                       if os.path.isfile(os.path.join(run_directory, f)) and not f.startswith('.')])
```

```python
# Exclusion list
exclude_strings = [
    'CanESM5', 'CNRM-CM6-1', 'EC-Earth3', 'IPSL-CM6A-LR',
    'UKESM1-0-LL', 'NorESM2-MM', 'EC-Earth3-CC', 'NESM3',
    'EC-Earth3-Veg', 'INM-CM4-8', 'TaiESM1', 'ssp585'
]
```

```python
# Dictionary to store valid matches
valid_matches = {}

# Process each file
for file in results_list:
    # Extract the number prefix (first two digits)
    index = int(file.split("_")[0])  # Convert to integer for indexing
    
    # Ensure index is within the directory list bounds
    if index < len(directories):
        directory = directories[index]  # Get corresponding directory
        
        # Check if directory contains any excluded strings
        if not any(exclude in directory for exclude in exclude_strings):
            valid_matches[file] = directory

            parts = directory.split('/')
            middle_part = '_'.join(parts[2:-2]).replace('/', '_')

            # Construct new filename
            base_name, ext = os.path.splitext(file)  # Get name and extension
            new_filename = f"{base_name}_{middle_part}{ext}"  # Append middle part

            # Get full paths
            old_path = os.path.join(run_directory, file)
            new_path = os.path.join(run_directory, new_filename)
        
            # Rename file
            os.rename(old_path, new_path)
            print(f"Renamed: {file} -> {new_filename}")
```
