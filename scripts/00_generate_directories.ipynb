```python
import os
```

```python
base_path = "../datatool/outs/"  # replace with your actual base directory path
directories = []

# Loop through the base directory and check each directory
for root, dirs, files in os.walk(base_path):
    # If the directory has no subdirectories, add it to the list
    if not dirs:  # dirs will be an empty list if no subdirectories exist
        directories.append(root)

# Print or save the list of full directory paths
print(directories)

# Optionally, save to a text file
with open("directories.txt", "w") as f:
    for directory in directories:
        f.write(f"{directory}\n")
```
