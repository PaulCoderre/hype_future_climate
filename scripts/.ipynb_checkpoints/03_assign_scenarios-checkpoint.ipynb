```python
import os
import glob
import shutil
```

### Inputs

```python
directory_list= './directories.txt' # path to datatool directories list
run_directory = './' # path to directory containing model run output folders
results_directory= './renamed_results/' # directory to save the updated filenames
```

### Reformat Files

```python
# Check if results_directory exists, if not, create it
if not os.path.exists(results_directory):
    os.makedirs(results_directory)
    print(f"Created directory: {results_directory}")
else:
    print(f"Directory already exists: {results_directory}")
```

```python
# Find all files matching run_directory/results_*/ and copy to results_directory, do this to move all results to 1 directory
pattern = os.path.join(run_directory, 'results_*', '*')
files_to_copy = glob.glob(pattern)
```

```python
len(files_to_copy)
```

```python
for file_path in files_to_copy:
    if os.path.isfile(file_path):  # Ensure it's a file
        shutil.copy(file_path, results_directory)
   #     print(f"Copied: {file_path} -> {results_directory}")

print("All matching files have been copied.")
```

```python
# Initialize an empty list to store the paths
directory_paths = []

# Open the file and read each line into the list
with open(directory_list, 'r') as file:
    for line in file:
        # Strip any leading/trailing whitespace characters
        path = line.strip()
        # Add the path to the list
        directory_paths.append(path)
```

```python
# Initialize a list to save formatted paths
formatted_paths = []

# Extract and format the last 5 fields
for path in directory_paths:
    formatted_path = '_'.join(path.split('/')[-5:])
    formatted_paths.append(formatted_path)
```

```python
# Iterate through each index number and string in found_strings
for index, found_string in enumerate(formatted_paths):
        # Iterate through files in results_directory
    for filename in os.listdir(results_directory):
        # Check if filename starts with the index (with leading zero)
        if filename.startswith(f"{index:02d}_"):
            # Append found string to the end of filename
            new_filename = os.path.join(results_directory, filename[:-4] + "_" + found_string + ".txt")
            os.rename(os.path.join(results_directory, filename), new_filename)
        #    print(f"Renamed file: {filename} to {new_filename}")
```

```python
# Locate and delete directories matching the patterns
patterns = [os.path.join(run_directory, 'results*'), os.path.join(run_directory, 'working_directory*')]

for pattern in patterns:
    # Get all matching directories
    matching_dirs = glob.glob(pattern)
    for directory in matching_dirs:
        # Remove the directory and its contents
        if os.path.isdir(directory):
            shutil.rmtree(directory)
            print(f"Deleted directory: {directory}")
```
