```python
import os
import glob
import shutil
```

```python
directory_list= './directories.txt' # path to datatool directories list
easymore_directory = '../region_easymore/' # path to directory containing model run output folders
results_directory= '../region_official_easymore/' # directory to save the updated filenames

# List of strings to exclude
exclude_strings = [
    'CanESM5', 'CNRM-CM6-1', 'EC-Earth3', 'IPSL-CM6A-LR',
    'UKESM1-0-LL', 'NorESM2-MM', 'EC-Earth3-CC', 'NESM3',
    'EC-Earth3-Veg', 'INM-CM4-8', 'TaiESM1', 'ssp585'
]
```

```python
# Read the list of paths from directories.txt
with open(directory_list, 'r') as f:
    directories = [line.strip() for line in f.readlines()]
```

```python
# Create results_directory if it does not exist
os.makedirs(results_directory, exist_ok=True)
```

```python
# Get list of folders in easymore_directory, sort alphabetically, and exclude hidden folders (starting with '.')
folders = sorted([f for f in os.listdir(easymore_directory) 
                  if os.path.isdir(os.path.join(easymore_directory, f)) and not f.startswith('.')])
```

```python
# Ensure the lists have the same length
if len(folders) != len(directories):
    print("Error: Number of folders and directory paths do not match!")
else:
    for idx, folder in enumerate(folders):
        original_path = directories[idx]  # Get corresponding path

        # Split the path and remove first 2 and last 2 elements
        parts = original_path.split('/')
        if len(parts) > 4:
            middle_part = '_'.join(parts[2:-2])  # Join middle part with '_'
        else:
            print(f"Skipping {folder} - not enough parts in path: {original_path}")
            continue
        
        # New folder name
        new_folder_name = f"{folder}_{middle_part}"
        old_path = os.path.join(easymore_directory, folder)
        new_path = os.path.join(easymore_directory, new_folder_name)

        # Rename the folder
        os.rename(old_path, new_path)
        print(f"Renamed: {folder} -> {new_folder_name}")

        # Check if folder should be excluded
        if any(ex_str in new_folder_name for ex_str in exclude_strings):
            print(f"Skipping {new_folder_name} (contains excluded string)")
            continue

        # Copy the folder to results_directory
        shutil.copytree(new_path, os.path.join(results_directory, new_folder_name))
        print(f"Copied {new_folder_name} to {results_directory}")
```
